```python

class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks  # List of 5 paper marks
        self.total = sum(marks)
        self.percentage = self.total / 5

    def check_admission(self):
        print("\n--- Student Details ---")
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")
        print(f"Total Marks: {self.total} / 500")
        print(f"Percentage: {self.percentage:.2f}%")

        if self.percentage >= 85:
            print("Status: Student can take admission")
        else:
            print("Status: Student can't take admission")
if __name__ == "__main__":
    name = input("Enter Student Name: ")
    roll_no = input("Enter Roll Number: ")
    print("Enter marks for 5 papers (out of 100):")
    marks = []
    for i in range(1, 6):
        m = float(input(f"Paper {i}: "))
        marks.append(m)
    student1 = Student(name, roll_no, marks)
    student1.check_admission()
```

```python
# Single level inheritance
class A:
    def showA(self):
        print("showA Function")

class B(A):  # Added (A) so class B inherits from class A
    def showB(self):
        print("showB function")

# Creating an object of the child class B
obj = B()
obj.showA()  # Now works because B inherits from A
obj.showB()
```

```python
class A:
    def showA(self):
        print("showA Function")
class B(A):
    def showB(self):
        print("showB function")
class C(B):
    def showC(self):
        print("showC function")
obj = C()
obj.showA()
obj.showB()
obj.showC()
```
