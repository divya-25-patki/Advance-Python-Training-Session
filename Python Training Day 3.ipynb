```python
import csv
with open("emp.csv","w",newline='') as f:
  w=csv.writer(f)
  w.writerow(["ENO","ENAME","ESALARY","EADDR"])
  n=int(input("Enter no. of employes : "))
  for i in range(n):
    eno=input("Enter employee no : ")
    ename=input("Enter employee name : ")
    esalary=input("Enter employee salary : ")
    eaddr=input("Enter employee address : ")
    w.writerow([eno,ename,esalary,eaddr])
print("Data inserted successfully!")
```

```python
import numpy as np
students = np.array([
    [5,9,7],
    [4,94,85],
    [5,0,81],
    [8,90,78],
    [95,98,8]
])
students
```

```python
avg_marks = np.mean(students,axis=1)
print("avg marks= ",avg_marks)
```

```python
avg_marks = np.mean(students,axis=0)
print("avg marks= ",avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("topper student index= ",topper)
```

```python
failed_student = np.where(avg_marks>50)
print(failed_student[0]+1)
```

```python
for i, avg in enumerate(avg_marks):
  status = "pass" if avg > 50 else "Fail"
  print(f"Student (i+i): avg=(avg:.2f)| (status)")
```

```python
months = ['jan','feb','mar','april']
sales = [250,389,496,400]
plt.bar(months,sales,color='blue')
```

```python
plt.barh(months,sales,color='orange')
plt.title('monthly sales')
plt.xlabel('sales')
plt.ylabel('months')
plt.show()
```

```python
height=[145,121,12,42,]
weight=[20,40,49,41]
plt.scatter(height,weight,color='red')
plt.title('height vs weight')
plt.xlabel('height')
plt.ylabel('weight')
plt.grid(True)
plt.show()
```

```python
brands = ['Brand A','Brand B','Brand C']
shares = [45,57,60]
plt.pie(shares,labels=brands,autopct='%1.1f%%', startangle=90)
plt.axis('equal')
plt.title('Market Shares')
plt.show()
```
